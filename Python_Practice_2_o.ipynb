# abstraction

```python
class Payment:
  def Pay(self):
    pass

class Paypal(Payment):
  def Pay(self):
    print(f"paid using Paypal")

class CreditCard(Payment):
  def Pay(self):
    print(f"paid using creditcard")
```

```python
method = Paypal()

method.Pay()
```

# encapsulation

```python
class Account:
  def __init__(self , balance):
    self.__balance = balance

  def show_balance(self):
    return self.__balance

class Transaction(Account):

  def deposit(self , amount):
    print(f"{amount} deposited")
    self._Account__balance += amount

  def withdrawl(self, amount):
    print(f"{amount} withdrawl")
    self._Account__balance -= amount




```

```python
tran = Transaction(70000)
```

```python
tran.deposit(1)
```

```python
tran.show_balance()
```
